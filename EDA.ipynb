### Distributions

```python
plot_cols = [col for col in features.columns if features[col].nunique() > 2]
num_rows = 11
num_columns = 6
# Create a figure with subplots
fig, axes = plt.subplots(num_rows, num_columns, figsize=(25, 6 * num_rows))  # Adjust figsize as needed
# Flatten axes array for easy iteration
axes = axes.flatten()
# Iterate over the columns and plot histograms
for i, col in enumerate(plot_cols):
    ax = axes[i]
    ax.hist(features[col], bins=30, edgecolor='k', alpha=0.7)
    ax.set_xlabel(col)
    ax.set_ylabel('Frequency')
    ax.grid(True)

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])
```

### Correlation Analysis

```python
plot_cols = [col for col in features.columns if features[col].nunique() > 2]
# Calculate the correlation matrix
correlation_matrix = features[plot_cols].corr()
# Plot the heat map
plt.figure(figsize=(80, 60))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heat Map')
plt.show()
```

### Feature vs Target Variable

```python
target_col = 'is_team1_winner'
plot_cols = [col for col in features.columns if (col!=target_col and features[col].nunique() > 2)]

num_rows = 11
num_columns = 9

# Create a figure with subplots
fig, axes = plt.subplots(num_rows, num_columns, figsize=(20, 6 * num_rows))  # Adjust figsize as needed

# Flatten axes array for easy iteration
axes = axes.flatten()

# Iterate over the columns and plot bar plots
for i, col in enumerate(plot_cols):
    ax = axes[i]
    # Define bins
    bins = np.linspace(features[col].min(), features[col].max(), 10)
    # Digitize the feature values into bins
    digitized = np.digitize(features[col], bins)
    # Calculate the count of 1s for each bin
    bin_counts = [features[target_col][digitized == j].sum() for j in range(1, len(bins))]
    # Plot the bar plot
    ax.bar(range(1, len(bins)), bin_counts, align='center', alpha=0.7, edgecolor='k')
    ax.set_xticks(range(1, len(bins)))
    ax.set_xticklabels([f"{bins[j-1]:.1f}-{bins[j]:.1f}" for j in range(1, len(bins))], rotation=45)
    ax.set_xlabel(col)
    ax.set_ylabel(f'Count of {target_col}=1')
    ax.grid(True)

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])
```

### Principal Components Plot

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import plotly.express as px
```

```python
pca_cols = [col for col in features.columns if (col!=target_col and features[col].nunique() > 2)]

# Standardize the data
scaler = StandardScaler()
features_standardized = scaler.fit_transform(features[pca_cols])

# Apply PCA
pca = PCA(n_components=3)  # Choose the number of components to keep
principal_components = pca.fit_transform(features_standardized)

# Create a DataFrame with the principal components
pc_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2', 'PC3'])
pc_df['target'] = features['is_team1_winner']

# Print explained variance ratio
explained_variance_ratio = pca.explained_variance_ratio_
print("Explained variance ratio:", explained_variance_ratio)
print("Total variance explained: ", sum(explained_variance_ratio))
print(" ")

# Create a DataFrame for explained variance ratio
variance_df = pd.DataFrame({
    'Principal Component': ['PC1', 'PC2', 'PC3'],
    'Explained Variance Ratio': explained_variance_ratio
})

# Plotting the explained variance ratio
variance_df.plot(kind='bar', x='Principal Component', y='Explained Variance Ratio', legend=False)
plt.ylabel('Explained Variance Ratio')
plt.title('Explained Variance Ratio of Principal Components')
plt.show()
```

```python
fig = px.scatter_3d(
    pc_df,
    x='PC1',
    y='PC2',
    z='PC3',
    color='target',
    # title='3D PCA Plot',
    labels={'PC1': 'PC1', 'PC2': 'PC2', 'PC3': 'PC3'},
    color_continuous_scale=px.colors.sequential.Viridis,
    # symbol='target',
    symbol_sequence=['circle', 'diamond'],
    template='plotly_dark'
)

# Update the layout for better visuals
fig.update_layout(
    scene = dict(
        xaxis_title='PC1',
        yaxis_title='PC2',
        zaxis_title='PC3',
        xaxis=dict(backgroundcolor="rgb(200, 200, 230)",
                   gridcolor="white",
                   showbackground=True,
                   zerolinecolor="white"),
        yaxis=dict(backgroundcolor="rgb(230, 200,230)",
                   gridcolor="white",
                   showbackground=True,
                   zerolinecolor="white"),
        zaxis=dict(backgroundcolor="rgb(230, 230,200)",
                   gridcolor="white",
                   showbackground=True,
                   zerolinecolor="white"),
    ),
    margin=dict(r=10, b=10, l=10, t=10),
    coloraxis_colorbar=dict(
        # title="Target",
        tickvals=[0, 1],
        ticktext=["Class 0", "Class 1"]
    )
)

fig.show()
```
